# Model Deployment
## Choosing the right model
Consider the trade-off between model interpretability and performance. There are models that provide interpretability at the level of total variance of the data, but also how each particular feature contributes to the total variance.

Always test multiple models and perform cross-validation in order to compare models fairly. Also perform tuning of hyperparameters. Use the same train/test split when comparing, as well as taking into account which data and models prefer to have scaled input.

"Choosing the right estimator" is a guide how to choose the adecuate algorithm for a particular case.

## Purpose of Deployment
The final approach as how to deploy a project depends on the final purpose of the model: it is different to construct a project portfolio to show from a enterprise deployment of a model for everyone to use.

### Project Portfolio
You may write a blog post (static, live from Git) instead of having a fully functional server with the project running 24/7. You may also set up a Flask-based website.

### Enterprise level deployment
If your project is to provide a required service for the enterprise, communicate with the stakeholders to find out what is the most suitable option.

## Performance expectations
Remember to keep a hold-out set of data to obtain metrics about non-seen data, which is what is really important.

1. Train/test/validation split
2. Hyperparameter tuning
3. Report results and metrics on the final holdout set
4. Retrain model on all data prior to deployment

## Retraining Intervals
If the performance is still good, no need.
If you suspect that the model's performance is worsening, try to see if it can be adjusted.

If you receive labeled data, try to include the new data to re-train.
