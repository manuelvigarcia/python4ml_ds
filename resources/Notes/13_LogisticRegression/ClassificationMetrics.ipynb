### Accuracy
Accuracy is how many predictions are correct out of the total. `(True possitive+true negative)/total`

**Accuracy paradox**

But this is not a good measure of the quality of the model, since many of the features in the real world are imbalanced and we try to predict something that is overwhelming tilted to one side. e.g. illnes vs. healthy, product fail vs. product perfect, natural dissaster vs. good living conditions. In those cases, a model that predicts always the majority class can be overwhelmingly accurate. e.g. if we are trying to predict the occurrence of a very rare illness with an occurrence of 1 case per 100.000 inhabitants, that is 0.00001%; if a model is always predicting healthy, it will be accurate 99.9999% of the cases; but it is not useful, since it fails at all the meaningful cases. Not going that far, in a situation where 5% of the population is ill, a model always predicting healthy would be 95% accurate, but not good at all, since it cannot identify the people that need treatment.

### Precission
Precission measures how many of the positive identifications are correct. i.e. When the model predicts positive, how often it is correct?

This is `the amount of true positives / total predicted positives(both true and false positives)`. In the case of the 5% population ill with a "always healthy" model, we have 0 predicted positives (it does not identify any of them) and a total of 0 true positives, leaving 0/0 indetermination.

### Recall
This measures how often is the prediction correct when it is actually a positive case. Of all the ill people, how often the model identified them?; of all the natural dissasters, how often the model blew the whistle?; of all the manufacturing errors, how ofthen the model idetified a failure?

This is calculated `true positives / total actual positives`. That is the amount of cases the model identified correctly as infected divided by the amount of actually infected people.

In the case of our 'always healthy' model, that would be 0 identified infected (it always says "not infected") / 5 actually infected people, resulting in a 0% recall.

### F1 score
`(2 x precision x recall) / (precision + recall)`

This way the score will be 0 if either precision or recall are 0.
