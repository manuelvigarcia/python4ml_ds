Having two arrays of the same dimension, you can order one and apply the same element-by-element sorting to the other; so the element originaly in pace nth from both arrays ends up in place mth in both arrays after sorting.

```python
import numpy as np
array1 = np.random.normal(1, 1, 80)
array2 = np.array(range(1,81))
#Let's say we have two arrays with two features of different objects. We want to sort the objects regarding feature 1, but then the other feature has to be sorted in the same order
indices=array1.argsort()  # gets the indices sorted according to the contents of array1
array1 = array1[indices]  # applies the sorted indices to array1
array2 = array2[indices]  # applies the sorted indices to array2
print(array1,'\n',array2,'\n',indices)
```
