We will work with the Iris Data Set: https://en.wikipedia.org/wiki/Iris_flower_data_set from a flower classification in 1936 with 3 classes of the same family of flowers.

```python
X = df.drop('species',axis=1)
y = df['species']
```

```python
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
```

```python
X_train, X_test, y_train,y_test = train_test_split(X,y,test_size=0.25, random_state=101)
```

```python
scaler = StandardScaler()
scaled_X_train=scaler.fit_transform(X_train)
scaled_X_test=scaler.transform(X_test)
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.model_selection import GridSearchCV
```

```python
log_model = LogisticRegression(solver='saga',multi_class='ovr', max_iter=5000)
```

```python
penalty=['l1','l2','elasticnet']
l1_ratio=np.linspace(0,1,20)
C=np.logspace(0,10,20)
param_grid={'penalty':penalty,'l1_ratio':l1_ratio,'C':C}
```

```python
grid_model = GridSearchCV(log_model,param_grid=param_grid)
```

```python
grid_model.fit(scaled_X_train,y_train)
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
from sklearn.metrics import ConfusionMatrixDisplay, confusion_matrix
```

```python
grid_model.best_params_
```

```python
y_pred=grid_model.predict(scaled_X_test)
y_pred
```

```python
accuracy_score(y_test,y_pred)
```

```python
cm=confusion_matrix(y_test,y_pred)
cm
```

```python
cm_display = ConfusionMatrixDisplay(cm).plot()
```

```python
# Or, directly from the model
ConfusionMatrixDisplay.from_estimator(grid_model,scaled_X_test,y_test)
```

Only one 'virginica' clasified as 'versicolor'.

```python
print(classification_report(y_test,y_pred))
```

## Plotting ROC curve for multiclass classification problem
The usual ROC curve plotting method does not work for multiclass problem.

You can work arround that by plotting each of the curves in a figure.

```python
from sklearn.metrics import PrecisionRecallDisplay
from sklearn.metrics import RocCurveDisplay
```
