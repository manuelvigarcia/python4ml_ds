# Cross Validation

The idea is to have some means to validate that the model we train is suitable to make predictions on the kind of data we get from the domain.

The most basic concept is to split the dataset available in two: one big part to train the model and one smaller part to assess that the model predicts correctly. In this smaller part we have the actual outcome that has happened, so we can compare what the model predicts for these data points with the real outcome that did actualy happen. The key idea is that the model must never see these data points until it is asked to predict the outcome; so these data points, the test set, shall be separated immediately after cleaning the dataset and before performing any taks to build the data model. Then the model is trained on the train portion of the data set. From time to time, the model is tried out with the test set and the resulting error is used to calibrate the model further.



### Train-Test Split Procedure
0. Clean and adjust data as necessary for X and y
1. Split Data in Train/Test for both X and y
2. Fit/Train Scaler on Training X Data
3. Scale X Test Data (unless data for every feature are in the same units)
4. Create Model
5. Fit/Train Model on X Train Data
6. Evaluate Model on X Test Data (by creating predictions and comparing to Y_test)
7. Adjust Parameters as Necessary and reapeat steps 5 & 6

Besides, having adjusted model_two with knowledge of how the model has performed with X_test data, means there is some knowledge about X_test that goes in building the model. This makes X_test loose the quality for which it was constructed: evaluate the model with some data completely unknown to the model.

Let's try to overcome this by using 'train-validation-test' split. It consists on holding a second test data set to validate the model at the end, when we are happy with its performance against the test set. That is the validation set.

## Train Validation Test split technique
- still use 70% for the train data set
- split the remaining 30 percent into two sets: one test set to use as the train-test split and another one for final validation of the model. The actual % depends on how much data is available. 10% for validation is a good percentage, if the amount of data is enough to make a fair validation of the model. If not, make the validation set bigger until you fill confident.
- The process goes as the previous technique until the model is satisfactory. Then run a final prediction against the validation data set and only report on the result, do not adjust hyperparameters, nor train the model any more; just make a report on the quality of the model against the validation set.

This will be done by splitting the data set twice. The first time take the training set. The remaining data are split into validation and test sets.

## Cross Validation
K-fold cross validation consists on:
- performing a split into train data and test data
- then the train data is split into K subsets (typical 5-10, with more data, you can use higher numbers)
- reserve one of the subsets for validation, train the model against the rest of the subsets and finally validate against the reserved one and calculate the error
- repeat reserving each of the K subsets in turns until the model has been tested against each of the K subsets. Then calculate the average of all K errors and use that to adjust the hyperparameters
- Repeat the whole K-split until you are satisfied with the resulting performance
- Use the test data to asses and report on the performance of the model.

cros_val_score in sklearn performs all this automatically.

```python
df = pd.read_csv('../../jupyter_notebooks/DATA/Advertising.csv')
df.head()
```

```python
X = df.drop('sales',axis=1)
```

```python
y = df['sales']
```

```python
X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=101)
```

```python
scaler = StandardScaler()
```

```python
scaler.fit(X_train)
```

```python
X_train = scaler.transform(X_train)
```

```python
X_test = scaler.transform(X_test)
```

```python
model = Ridge(alpha=100)
```

```python
from sklearn.model_selection import cross_val_score
```

Now we have to choose a scoring metric. It can be chosen from https://scikit-learn.org/stable/modules/model_evaluation.html

```python
scores = cross_val_score(model,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

The scores calculated by cross_val_score are all arranged so "higher is better"; in the case of the meann squared error, is the negative of those values, so to make it signify "better" when it is "higher".

In order to be able to compare this with the performance we had before, we'll have to use the absolute value.

```python
scores
```

These are 

```python
abs(scores.mean())
```

Not very good performance. Lets make another model with alpha= 1.

```python
model = Ridge(alpha=1)
```

```python
scores = cross_val_score(model,X_train,y_train,scoring='neg_mean_squared_error',cv=5)
```

```python
scores
```

```python
abs(scores.mean())
```

The model is only fitted inside cross_val_score. So in order to perform the final evaluation, we have to fit the model ourselves with the train data.

```python
model.fit(X_train,y_train)
```

```python
y_final_test_pred=model.predict(X_test)
```

```python
mean_squared_error(y_test,y_final_test_pred)
```

And this is the value you may use as reported expected performance of the model we've just trained.

## cross_validate function to perform different metrics on the fitted model performance
https://scikit-learn.org/stable/modules/model_evaluation.html

```python
X = df.drop('sales',axis=1)
y = df['sales']
```

```python
X_train,X_test,y_train,y_test = train_test_split(X,y,test_size=0.3,random_state=101)
```

```python
scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
from sklearn.model_selection import cross_validate
```

```python
model = Ridge(alpha=100)
```

```python
scores = cross_validate(model,X_train,y_train, scoring=['neg_mean_squared_error', 'neg_mean_absolute_error'],cv=10)
```

```python
scores = pd.DataFrame(scores)
```

```python
scores = pd.DataFrame(scores)
```
