```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
df = pd.read_csv('../../jupyter_notebooks/08-Linear-Regression-Models/Advertising.csv')
df
```

```python
df.head()
```

```python
df['total_spent'] = df['TV']+df['radio']+df['newspaper']
df.head()
```

```python
sns.scatterplot(data=df, x='total_spent',y='sales')
```

```python
# scatterplot + regression line
sns.regplot(data=df, x='total_spent',y='sales')
```

```python
X=df['total_spent']
y=df['sales']
```

```python
#help(np.polyfit)
# y = B1x + B0
coefficients= np.polyfit(X,y,deg=1)
coefficients
```

```python
potential_spend = np.linspace(0,500,100)
```

```python
predicted_sales = coefficients[0] * potential_spend + coefficients[1]
```

```python
#Regression line calculated by hand
plt.plot(potential_spend,predicted_sales)
```

```python
# Add the data scatter plot to the regression line
plt.plot(potential_spend,predicted_sales,color='red')
sns.scatterplot(data=df,x='total_spent',y='sales')
```

```python
# now, instead of a lot of possible points, we try the 'what if we spend 200 K$ in advertising, what sales can we expect?'
spend = 200
predicted_sales = coefficients[0] * spend + coefficients[1]
predicted_sales
```

```python
#higher degree polinomial: 3rd degree
#  y = B3x^3 + B2x^2 + B1x^1 + B0x^0 = B3x^3 + B2x^2 + B1x + B0
coefficients_3 = np.polyfit(X,y,deg=3)
coefficients_3
```

```python
# let's plot the 3rd degree regression line
pot_spend = np.linspace(0,500,100)
pred_sales = coefficients_3[0] * pot_spend**3 +  coefficients_3[1] * pot_spend**2 + coefficients_3[2] * pot_spend + coefficients_3[3]
```

```python
plt.plot(pot_spend,pred_sales,color='red')
sns.scatterplot(data=df,x='total_spent',y='sales')
```
