# Feature Engineering
Feature selection and data cleaning.

## Dates
These are usually dates as one data point. For them to be used by a model, they shall be integers or floating point numbers. Decomposing them into three separate features (Year, Month, Day) works, having a timestamp (in the way "Seconds since era start") also works, but makes it more obscure, since it cannot be easily interpreted by humans. Week days may be converted into 1-7 one digit data points.

## Text
Text fields can be processed to have them in int/float form by looking into the meaning or how many times some keywords are present there.

## Combining information
Already done with polynomial regression.

Can be done by grouping features that have synergies, or multiply features which, being desirable, don't usually come together, etc.

Extract information, like creating a new feature containing whether a business opens during weekends and evenings, so it would have "Nightlife" active.

## Transforming information
Common for string data.


### Integer Enconding
**Integer encoding** uses natural numbers to encode each of the possible values (categories) of the given feature. You can also choose a known algorithm to encode the feature, like encoding country names with the phone national prefix, or colours with their RGB (or other color space) value. This introduces the problem of "ordering" given by the numbers themselves: the model might understand that the third encoded value is worth three times the first encoded value, or some precedence among them...

You have to keep in mind any actual order among the categories, to encode them making the order explicit, like e.g. the priority of a task, or the seriousness of a medical condition.

### One Hot Encoding
**One Hot Encoding (Dummy variables)** uses one new feature for each existing category, so the corresponding feature will have value **1** when the original feature holds the category that is equal to this new feature and **0** in all other cases. So for a feature that can take three values A,B,C, there will be three new features called A, B, and C; Feature A will be 1 in the original feature had the value A for this observation and 0 otherwise; Feature B will be 1 if the original feature has value B for this observation and 0 otherwise; and the same for new feature C (1 only if original was C). Be careful to have one new feature for every possible value of the original feature. One shortcomming of this technique is that it increases the number of features of the dataset, but it avoids implying any order or precedence among values.

In some cases the number of possible categories (and hence the new features) may be too big, so you may want to transform the categories in higher level categories. For example going for continents or subcontinents instead of countries; for quarters instead of months; for instrument families instead of instruments; Genus, Families or Orders instead of species. . .

This can be easily done with pandas _.map()_ or _.apply()_ and when going for higher level categories, it requires domain expertise.

One more problem of Dummy Variables is the multicolinearity: one column depends on another or all other. For example, encoding direction what is not left is right, so only one is needed; if you have the physical state you only need 'gas', 'plasma', and 'liquid', because everything else is 'solid'; in the case of five possible categories, the fifth one is what rests when all other four are zero. So you can just drop one of the columns, because its value is encoded in the set of all other columns. Pandas has included the parameters "drop_first" and "drop_last" to account for this. The shortcomming of dealing with multicolinearity is that if in the future one more category appears in the original feature, you must add a new Dummy Variable with all previous observations with value zero and one in the new observation; example of this might be when shipping products to a new market/country, accounting for a new medical condition, dealing with a new product (e.g. hybrid fruit).

## Challenges in Feature Engineering (and data cleaning)
- Outliers
- Missing data

There is no "one fits all".

## Identifying the features that are worth the effort
You want clean and nice values in the features that will influence most in the prediction of the prediction label. To identify those, look for high correlation with the prediction label. Take care of cleaning nicely the features that are most correlated with the prediction label and the rest you can wing along, like using averages for the NA's, reconverting outliers to averages, etc. instead of removing outliers for every one of the features, which would result in considerable higher loss of observations.

But use always common sense. If one observation is outlier across many features, it is probably something you have to investigate (the obvious choice is letting it go, but it has to be checked if it conveys more information than just being an outlier: it may well be an entry error, but it might be also the first of a newer trend that has just started. . .)


## Outliers
We will be looking for data points that are outside the data trend we want to model, for which we want to predict. But not too many: should not be a representative percentage of the sample. If you identify too many outliers, that means your data distribution is very wide and you should use modeling techniques adapted to that fact.

Knowing what an outlier is requires domain knowledge. One particular case is changing trends: what is now an outlier might be the future tendency and become normal in the near future.

Also, removing outliers poses restrictions to the model: removing products that are "too expensive" means that our model will not be able to predict "too expensive" prices with accuracy; removing people outside "the age fork of interest" means that our model will not be accurate for people outside "the age fork of interest". As a golden rule, if you put a limit to the data you use to build your model, you should maintain that limit for the data points you will try to predict with your model.

There is no 100% correct methodology that will apply to every situation.

### Methods for identifying Outliers
- **InterQuartile Range**. The interquartile range is IQR = q3-q1. It is common to use this IQR to identify outliers by assuming that any data point higher than q3 + 1.5 * IQR is an outlier and any data point below q1 - 1.5 * IQR is also an outlier. You may choose to use a different coeficient than 1.5 in order to tune the amount of outliers to drop.
- **Standard Deviation**. For data that follows a Normal distribution, 3 * Standard Deviation is a common measure to identify outliers: any data point further away than 3 * StDev from the mean value is an outlier. Again, you might change the "3" factor to tune how many outliers you leave out.
- **Visualization**. You may plot your chosen feature against the label you want to predict and visually identify the values that fall way outside the trend you want to model. Then you can just identify the data points by their values and leave them outside.
- **Domain Limit Value**. If you counting with domain knowledge (yours or the advice of an expert) you might identify a value limit ouside which the model is allowed to have less accuracy, then you leave out all data points with a value outside that limit. For example: cars with more than 500 horsepower, Recreational vehicles over 100.000€, houses over 10.000.000€
