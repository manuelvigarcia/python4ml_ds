Scikit-learn's GridSearchCV is capable of runing several combinations of hyperparameter values to find the best combination to attain the best performance of the model.

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error
from sklearn.preprocessing import StandardScaler
from sklearn.linear_model import ElasticNet
from sklearn.model_selection import GridSearchCV
```

```python
df = pd.read_csv('../../jupyter_notebooks/DATA/Advertising.csv')
df.head()
```

```python
X = df.drop('sales',axis=1)
y = df['sales']

X_train, X_test, y_train, y_test = train_test_split(X,y,test_size=0.3, random_state=101)

scaler = StandardScaler()
scaler.fit(X_train)
X_train = scaler.transform(X_train)
X_test = scaler.transform(X_test)
```

```python
base_elastic_net_model = ElasticNet()
```

```python
param_grid={'alpha':[0.1,1,5,10,50,100],'l1_ratio':[.1,.5,.7,.95,.99,1]}
```

```python
grid_model=GridSearchCV(estimator = base_elastic_net_model,param_grid = param_grid, scoring='neg_mean_squared_error',cv=5,verbose=2)
```

```python
grid_model.fit(X_train,y_train)
```

```python
grid_model.best_estimator_
```

```python
grid_model.best_params_
```

```python
pd.DataFrame(grid_model.cv_results_)
```

```python
y_pred = grid_model.predict(X_test)
```

```python
mean_squared_error(y_test,y_pred)
```
