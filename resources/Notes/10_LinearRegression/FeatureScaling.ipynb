# Feature scaling

### Standardization

Rescales to a mean of 0 and a Standard Deviation of 1

```python
rescaled = (x-x.mean)/x.stdev
```

### Normalization

rescales all values to the range (0,1)

```python
normalized = (x-x.min)/(x.max-x.min)
```

## What to scale

Scale all data.

Do not scale the "label", "y", prediction. You want the predicted data to still be in the real world "idiom".

Scaling the label can actually harm stochastic gradient descent.

## How to scale

.fit() method will calculate the necessary statistics to later perform the rescaling

Only call .fit() to the training data. If you use the entire data set, you are leaking test data information into the model. You want the test data to be completely unknown to the model.

.transfrom() method will actually rescale the data. You shall transform the whole data set: train and test.

* Split into train and test
* Fit to the training data
* Transform the training data
* Train the model
* Transform the test data
* Test the model with the transformed test data
